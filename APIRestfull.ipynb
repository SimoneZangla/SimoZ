#**ApiRestfull**

API sta per "Application Programming Interface" ed è una raccolta di metodi e
funzioni
implementati in un sistema software in modo che sia possibile comunicare con altre applicazioni.
In altre parole, un'API consente l'utilizzo di un servizio (funzione o metodo) di una piattaforma a
un sistema esterno, affinché possa essere utilizzato in un'applicazione proprietaria. Tali "metodi e
funzioni" possono includere operazioni di creazione, lettura, aggiornamento ed eliminazione. Un esempio calzante può essere il sito Trivago, in quanto selezionando il luogo e la data di interesse e facendo clic su "Cerca", il sito interagisce con l'API di ciascun hotel, fornendo
risultati per le camere disponibili che soddisfano i tuoi criteri.

# Che cos'è REST?

Nel 2000, Roy Fielding ha introdotto il concetto di REST (Representational State Transfer)
definendolo come uno stile architetturale per la progettazione e la creazione di servizi Web, in
altre parole REST rappresenta un insieme di regole che definiscono come scambiare risorse in un
sistema distribuito
I servizi web che implementano queste regole sono definiti RESTful.
Vediamo quali sono queste regole:
1. **Client-server:** la comunicazione tra le applicazioni segue il paradigma c/s. Questo
disaccoppiamento garantisce che, purché sia stata concordata un'interfaccia, lo sviluppo di
client e server può essere eseguito in modo indipendente;
2. **Stateless:** Nessuno stato client è memorizzato sul server. Tutte le informazioni necessarie per
eseguire le operazioni sono contenute nelle richieste;
3. **Cacheable:** I servizi Web RESTful devono fornire funzionalità di memorizzazione nella cache. I
server possono indicare come e per quanto tempo memorizzare nella cache le risposte; I
client possono utilizzare le risposte memorizzate nella cache invece di contattare il server.
Poiché REST in genere utilizza HTTP, eredita tutte le proprietà di memorizzazione nella cache
offerte da HTTP.
4. **Interfaccia uniforme:** le risorse che il server espone devono essere identificate da URI;
5. **Sistema a strati:** Data l’architettura c/s, i client non sono consapevoli del server specifico con
cui interagiscono. Questa proprietà consente l'introduzione di server intermedi che possono,
ad esempio, gestire la sicurezza o offrire funzionalità di bilanciamento del carico;
6. **Codice su richiesta (opzionale):** Anche se fa parte dell'architettura REST, questo principio è
facoltativo. I server possono estendere le funzionalità dei client trasferendo codice eseguibile.
REST, per definizione, è indipendente da qualsiasi protocollo sottostante e non è
necessariamente collegato a http, tuttavia le implementazioni REST più comuni usano HTTP
come protocollo applicativo.

# Che cos'è un RESTfull WebService?

Quando si sono diffuse le API per applicazioni e piattaforme Web, è stato introdotto il termine
RESTFUL WEB SERVICE, che viene utilizzato per denominare quelle API implementate con il
protocollo HTTP che forniscono una mappatura tra i verbi HTTP (get, post, put, delete) e le
azioni CRUD (create, read, update, delete).

**API vs Web Service**

API e Web Service sono due sistemi di comunicazione tra applicazioni. La principale differenza è
che un'API funge da interfaccia tra due diverse applicazioni in modo che possano comunicare tra
loro mentre un Web service ha lo stesso compito ma le applicazioni che mette in comunicazione
sono in rete. Un’API può utilizzare qualsiasi mezzo di comunicazione per avviare l'interazione tra
le applicazioni (es. API del sistema operativo sono utilizzabili chiamando delle funzioni in un
programma software). Un servizio Web usa http.
Tutti i WebService sono API ma tutte le API non sono WebService.
Le api che seguono i principi di rest e lavorano su http sono chiamate Restful API o Restful web
services

Uno dei vantaggi principali di REST su HTTP è costituito dal fatto che usa standard aperti e non
vincola l'implementazione dell'API o le applicazioni client a un'implementazione
specifica. Qualsiasi client deve poter chiamare l'API indipendentemente da come è implementata
internamente. Ad esempio, un servizio Web RESTful scritto in PHP, può essere consumato da
un’applicazione client scritta in un qualsiasi linguaggio (Java ad esempio) che può generare
richieste HTTP e analizzare risposte HTTP.
Vediamo i principali principi di progettazione di restful api che usano HTTP:


*   Le api sono progettate in base a risorse, che possono essere costituite da
qualsiasi
tipologia di oggetto, servizio o dato accessibile dal client.
*   Una risorsa ha un identificatore, costituito da un URI che identifica in modo univoco la
risorsa stessa.
*   I client interagiscono con un servizio scambiando rappresentazioni delle risorse. Molte
API usano JSON come formato di scambio.
*   Le API usano un'interfaccia uniforme che consente di separare le implementazioni del
client e del servizio. Per le API basate su HTTP, l'interfaccia uniforme include l'uso di verbi
HTTP standard per l'esecuzione di operazioni sulle risorse. Le operazioni più comuni sono
GET, POST, PUT, PATCH e DELETE.


*   Le API usano un modello di richiesta senza stato. Le richieste HTTP devono essere
indipendenti e possono essere eseguite in qualsiasi ordine, di conseguenza non è
possibile mantenere informazioni temporanee sullo stato tra le richieste. L'unico punto in
cui sono memorizzate le informazioni sono le risorse stesse e ogni richiesta deve essere
un'operazione atomica. Questo vincolo garantisce ai servizi Web un'elevata scalabilità,
perché non è necessario mantenere alcuna affinità tra i client e server specifici. Qualsiasi
server può gestire qualsiasi richiesta proveniente da qualsiasi client.
*   Le API REST si basano su collegamenti ipermediali contenuti nella rappresentazione
(HATEAOS*). Ad esempio nella rappresentazione JSON di un ordine sono contenuti i
collegamenti per recuperare o aggiornare il cliente associato all'ordine.

# URI per le risorse

Le risorse sono tipicamente dei contenuti che vogliamo ottenere o modificare da una
piattaforma Web tramite la sua API REST. Saranno identificate da un URI, che ci aiuterà a
ottenerli / modificarli.
Gli URI devono soddisfare determinate caratteristiche:

*   Non devono contenere un verbo che implichi un'azione

Errato : / resources / id / edit

Corretto : / resources / id /

*   Devono identificare una sola risorsa cioè devono essere univoci.

*   Non dovrebbero tenere conto del formato nella costruzione dell'URI

Errato : /resources /archivo.pdf

Corretto : / resources / file

*   Devono mantenere una gerarchia logica.

Errato : / appartamenti / 2 / edificio / 3

Corretto : / edifici / 3 / appartamenti / 2

*   Il filtraggio delle informazioni deve essere eseguito utilizzando i parametri HTTP

Errato : / edifici / 3 / appartamenti / 2 / ordine / asc

Corretto : / edifici / 3 / appartamenti / 2 /? Ordine = asc

*   Devono essere specificati usando la loro forma plurale

Errato : / appartamento / 2 /

Corretto : / appartamenti / 2 /

# USO dei metodi HTTP

Lo scopo di REST è modellare entità e le operazioni che possono essere eseguite da
un'applicazione su tali entità utilizzando i metodi HTTP.
Il protocollo HTTP definisce una serie di metodi che assegnano un significato semantico a una
richiesta. I metodi HTTP sono:

*  GET recupera una rappresentazione della risorsa nell'URI specificato. Il corpo del messaggio di risposta contiene i dettagli della risorsa richiesta;

*   POST crea una nuova risorsa nell'URI specificato. Il corpo del messaggio di richiesta fornisce i dettagli della nuova risorsa;

*   PUT crea o sostituisce la risorsa nell'URI specificato. Il corpo del messaggio di richiesta specifica la risorsa da creare o aggiornare;

*   PATCH esegue un aggiornamento parziale di una risorsa. Il corpo della richiesta specifica il set di modifiche da applicare alla risorsa;

*   DELETE rimuove la risorsa nell'URI specificato.
L'effetto di una specifica richiesta varia a seconda che la risorsa sia una raccolta o un singolo elemento.

# Considerazioni sulla progettazione delle API

Nell’identificazione delle risorse da esporre attraverso l’API bisogna tener presente che non è
necessario che una risorsa sia basata su un singolo elemento di dati fisico: ad esempio, una
risorsa ordine potrebbe essere implementata internamente in diverse tabelle di un database
relazionale, ma essere presentata al client come una singola entità.

Le entità sono spesso raggruppate in raccolte (ad esempio, ordini e clienti). Una raccolta è una
risorsa separata dall'elemento al suo interno e deve avere un proprio URI.
Inviando una richiesta HTTP GET all'URI della raccolta viene recuperato un elenco degli elementi
al suo interno. Ogni elemento della raccolta, a sua volta, ha un proprio URI univoco.

Una richiesta HTTP GET all'URI di un elemento restituisce i dettagli di tale elemento.

Un altro fattore da tenere in considerazione nella progettazione degli URI è che tutte le richieste
Web impongono un carico sul server Web. Maggiore è il numero di richieste, maggiore è il
carico. Vanno evitate quindi API Web "frammentate" che espongono un numero elevato di
risorse di piccole dimensioni. Un'API di questo tipo può richiedere che un'applicazione client invii
più richieste per trovare tutti i dati necessari. Potrebbe essere invece opportuno denormalizzare i
dati e combinare le informazioni correlate in risorse di maggiori dimensioni recuperabili con una
singola richiesta. Questo approccio deve tuttavia essere bilanciato rispetto al sovraccarico
associato al recupero di dati non necessari al client. Il recupero di oggetti di grandi dimensioni
può aumentare la latenza di una richiesta e comportare costi aggiuntivi in termini di larghezza di
banda.

# Rappresentazione di una risorsa

Client e server si scambiano rappresentazioni delle risorse. In una richiesta POST, ad esempio, il
corpo della richiesta contiene una rappresentazione della risorsa da creare. In una richiesta GET,
il corpo della risposta contiene una rappresentazione della risorsa recuperata.

Nel protocollo HTTP, i formati vengono specificati usando tipi di supporto, denominati anche tipi
MIME. Per i dati non binari, la maggior parte delle API Web supporta JSON (ovvero il tipo di
supporto application/json) ed eventualmente XML (ossia il tipo di supporto application/xml).

L'intestazione Content-Type in una richiesta o una risposta specifica il formato della
rappresentazione. Se il server non supporta il tipo di supporto, deve restituire il codice di stato
HTTP 415 (Tipo di supporto non supportato).

Una richiesta client può includere un'intestazione Accept contenente un elenco dei tipi di
supporto accettati dal client nel messaggio di risposta proveniente dal server. Se il server non
può fornire un tipo di supporto corrispondente a quelli elencati, deve restituire il codice di stato
HTTP 406 (Pagina non valida).

# Codici di stato

**Metodi GET**

Un metodo GET riuscito restituisce in genere il codice di stato HTTP 200 (OK). Se la risorsa non è
stata trovata, il metodo deve restituire 404 (Non trovato).

**Metodi POST**

Se un metodo POST crea una nuova risorsa, restituisce il codice di stato HTTP 201 (Creato). L'URI
della nuova risorsa è incluso nell'intestazione Location della risposta. Il corpo della risposta
contiene una rappresentazione della risorsa.
Se esegue alcune attività di elaborazione ma non crea una nuova risorsa, il metodo può restituire
il codice di stato HTTP 200 e includere il risultato dell'operazione nel corpo della risposta. In
alternativa, se non sono presenti risultati, il metodo può restituire il codice di stato HTTP 204
(Nessun contenuto) senza corpo della risposta.
Se il client inserisce dati non validi nella richiesta, il server deve restituire il codice di stato HTTP
400 (Richiesta non valida). Il corpo della risposta può contenere informazioni aggiuntive
sull'errore o un collegamento a un URI che offre maggiori dettagli.

**Metodi PUT**

Se un metodo PUT crea una nuova risorsa, restituisce il codice di stato HTTP 201 (Creato), così
come un metodo POST. Se il metodo aggiorna una risorsa esistente, restituisce 200 (OK) o 204
(Nessun contenuto). In alcuni casi potrebbe non essere possibile aggiornare una risorsa
esistente. In tale circostanza, valutare la possibilità di restituire il codice di stato HTTP 409
(Conflitto).

**Metodi PATCH**

Con una richiesta PATCH, il client invia un set di aggiornamenti a una risorsa esistente, sotto
forma di documento di patch. Il server elabora il documento di patch per eseguire
l'aggiornamento. Il documento di patch non descrive l'intera risorsa, ma solo un set di modifiche
da applicare. La specifica per il metodo PATCH (RFC 5789) non definisce un particolare formato
per i documenti di patch. Il formato deve essere dedotto dal tipo di supporto nella richiesta.

Il formato di dati più comune per le API Web è probabilmente JSON. Esistono due formati
principali di patch basati su JSON, denominati patch JSON e patch JSON di tipo merge.

Nella patch JSON di tipo merge il documento di patch presenta la stessa struttura della risorsa
JSON originale, ma include solo il subset di campi da modificare o aggiungere. È anche possibile
eliminare un campo specificando null come valore del campo nel documento di patch. La patch
di tipo merge non è quindi adatta se la risorsa originale può contenere valori Null espliciti.

# Filtrare e impaginare i dati

Esponendo una raccolta di risorse tramite un singolo URI, le applicazioni potrebbero recuperare
grandi quantità di dati quando è necessario solo un subset delle informazioni. Si supponga ad
esempio che un'applicazione client debba trovare tutti gli ordini con un costo superiore a un
valore specificato. Potrebbe recuperare tutti gli ordini dall'URI /orders e quindi filtrare tali ordini
sul lato client. Questo processo è chiaramente molto inefficiente, perché comporta uno spreco di
larghezza di banda di rete e di potenza di elaborazione nel server che ospita l'API Web.

L'API può invece consentire di passare un filtro nella stringa di query dell'URI, ad
esempio /orders?minCost=n. L'API Web sarà quindi responsabile dell'analisi e della gestione del
parametro minCost nella stringa di query e della restituzione dei risultati filtrati sul lato server.

Un altro aspetto da considerare è quando le richieste GET su risorse raccolta potrebbero
restituire un numero elevato di elementi. È consigliabile progettare un'API Web in modo da
limitare la quantità di dati restituita da ogni singola richiesta e quindi valutare la possibilità di
supportare stringhe di query che specificano il numero massimo di elementi da recuperare e un
offset iniziale nella raccolta. ad esempio: /orders?limit=25&offset=50
