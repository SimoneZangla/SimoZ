# **URI mapping htaccess**


**Redirect degli url**

 Quando progettiamo un’api dobbiamo:

● identificare gli endpoint delle risorse che vogliamo esporre;

● Identificare le operazioni che vogliamo consentire su quelle risorse individuando il metodo http idoneo.

Il client della nostra api potrà quindi costruire una richiesta http
verso la nostra api:

 Es. GET localhost/rest_api/basi_spaziali

 GET localhost/rest_api/basi_spaziali/6

 GET localhost/rest_api/basi_spaziali/77

**URI mapping**

 Se dovessimo mappare in modo identico l’alberatura di file e
directory che le uri ci suggeriscono avremmo una situazione
complessa (tante directory e tanti duplicati) e in continua
evoluzione (se aggiungo missioni?).

Come fare?

La soluzione è di inoltrare tutte le richieste verso l’applicazione
rest_api verso uno (o più) script che siano in grado di :

● comprendere l’uri di richiesta;

● fornire il modo dinamico la (rappresentazione) risorsa corretta.

 Nel nostro caso (XAMPP) utilizzeremo il modulo mod_rewrite
 del server web apache.

# File  .htaccess

Un file .htaccess è un file di testo che contiene delle regole di
comportamento a cui il server web deve attenersi quando riceve
una richiesta http. E’ possibile, ad esempio, scrivere regole per:

● Reindirizzare URL;

● Limitare o bloccare gli accessi ad alcuni file o directory;

● Limitare o bloccare gli accessi ad alcuni utenti;

● Gestire errori.

Il file .htaccess opera a livello della directory in cui si trova e
influenza il comportamento del server quando riceve una richiesta indirizzata verso quella directory


**Mod rewrite**

Per reindirizzare le nostre richieste dobbiamo:

 ● Creare un file .htaccess nella root directory della nostra applicazione web. (Negli esempi sopra ho utilizzato rest_api);

 ● Scrivere la direttiva:

RewriteEngine On

 ● Elencare di seguito le regole RewriteRule di redirect. Il
formato per specificare ogni regola è:

 RewriteRule EXPR URL [FLAG,FLAG]

**Rewrite rule**

EXPR -> Espressione regolare per indentificare una url di
richiesta.

 ● Ci sono alcuni caratteri jolly con un significato speciale:

Screenshot 2025-05-05 113041.png

Screenshot 2025-05-05 113123.png

URL -> Url reale del file a cui la richiesta deve essere
reindirizzata.

● Per parametrizzare la nuova richiesta oltre ad indicare il nome dello script da richiamare possiamo inserire parti dell’espressione regolare;

● Le parti che vogliamo utilizzare si devono devono essere messe tra parentesi tonde nell’espressione regolare;

● Per recuperarle si usa il dollaro “$” seguito da una cifra numerica ad indicare la posizione di questa corrispondenza.


FLAG -> Flag che modificano il funzionamento del mod_rewrite.

● Noi utilizzeremo:

○ QSA –querystringappend: aggiunge all’URL reale la query string presente al livello della finta URL. Utile per portarsi dietro contenuti appesi in query string o anche variabili di cui non avevamo tenuto conto;

○ NC –no case: valuta l’espressione regolare senza tenere conto di maiuscole o minuscole.
