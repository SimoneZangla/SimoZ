AJAX è il sogno di ogni sviluppatore, perché consente di:
* Leggere i dati da un server web - dopo che la pagina è stata caricata;

# Che cosa è AJAX?

AJAX = Asynchronous JavaScript And XML.

AJAX non è un linguaggio di programmazione.
AJAX utilizza semplicemente una combinazione di:
1. Un oggetto integrato nel browser XMLHttpRequest(per richiedere dati da un server web);
2. JavaScript e HTML DOM (per visualizzare o utilizzare i dati).

Le applicazioni AJAX potrebbero usare XML per trasportare dati, ma è altrettanto comune trasportare dati come testo normale o testo JSON.

AJAX consente alle pagine web di essere aggiornate in modo asincrono scambiando dati con un server web dietro le quinte. Ciò significa che è possibile aggiornare parti di una pagina web, senza ricaricare l'intera pagina.

# L'oggetto XMLHttpRequest

Tutti i browser moderni supportano l' XMLHttpRequestoggetto.

L' XMLHttpRequestoggetto può essere utilizzato per scambiare dati con un server web dietro le quinte. Ciò significa che è possibile aggiornare parti di una pagina web, senza ricaricare l'intera pagina.

**COME CREARE UN OGGETTO XMLHttpRequest**

Tutti i browser moderni (Chrome, Firefox, IE, Edge, Safari, Opera) hanno un oggetto integrato XMLHttpRequest.

Sintassi per la creazione di un XMLHttpRequestoggetto:

```python
variable = new XMLHttpRequest();
```

**COME DEFINIRE UNA FUNZIONE DI CALLBACK**

Una funzione di callback è una funzione passata come parametro a un'altra funzione.

In questo caso, la funzione di callback dovrebbe contenere il codice da eseguire quando la risposta è pronta:

```python
xhttp.onload = function() {
  // What to do when the response is ready
}
```

**COME INVIARE UNA RICIESTA**

Per inviare una richiesta a un server, puoi utilizzare i metodi open() e send() dell'oggetto XMLHttpRequest:

```python
xhttp.open("GET", "ajax_info.txt");
xhttp.send();
```

# Richieste XMLHttp

Il parametro url del open() metodo è un indirizzo di un file su un server:

```python
xhttp.open("GET", "ajax_test.asp", true);
```

Il file può essere di qualsiasi tipo, come .txt e .xml, oppure file di scripting del server come .asp e .php (che possono eseguire azioni sul server prima di inviare la risposta).

**ASINCRONO: VERO O FALSO?**

Le richieste al server devono essere inviate in modo asincrono.

Il parametro async del metodo open() dovrebbe essere impostato su true.

Inviando in modo asincrono, JavaScript non deve attendere la risposta del server, ma può invece:

* eseguire altri script in attesa della risposta del server;
* gestire la risposta dopo che la risposta è pronta.

Il valore predefinito per il parametro async è async = true.

Puoi tranquillamente rimuovere il terzo parametro dal tuo codice.

Synchronous XMLHttpRequest (async = false) non è consigliato perché JavaScript smetterà di essere eseguito finché la risposta del server non sarà pronta. Se il server è occupato o lento, l'applicazione si bloccherà o si arresterà.



**GET o POST?**

GETè più semplice e veloce di POSTe può essere utilizzato nella maggior parte dei casi.

Tuttavia, utilizzare sempre le richieste POST quando:

* Un file memorizzato nella cache non è un'opzione (aggiornare un file o un database sul server);
* Invio di una grande quantità di dati al server (POST non ha limiti di dimensione);
* Inviando input utente (che possono contenere caratteri sconosciuti), POST è più robusto e sicuro di GET.

**ES. UTILIZZO DI RICHIESTE GET**

Se vuoi inviare informazioni con il GETmetodo, aggiungi le informazioni all'URL:

```python
xhttp.open("GET", "demo_get2.asp?fname=Henry&lname=Ford");
xhttp.send();
```

**ES. UTILIZZO DI RICHIESTE POST**

Per inviare dati come un modulo HTML, aggiungi un'intestazione HTTP con setRequestHeader(). Specifica i dati che vuoi inviare nel send()metodo:

```python
xhttp.open("POST", "ajax_test.asp");
xhttp.setRequestHeader("Content-type", "application/x-www-form-urlencoded");
xhttp.send("fname=Henry&lname=Ford");
```

**RICHIESTA SINCRONA**

Per eseguire una richiesta sincrona, modificare il terzo parametro nel open()metodo in false:

```python
xhttp.open("GET", "ajax_info.txt", false);
```

A volte async = false vengono usati per test rapidi. Troverai anche richieste sincrone nel codice JavaScript più vecchio.

Poiché il codice attenderà il completamento del server, non è necessaria alcuna onreadystatechange funzione:

```python
xhttp.open("GET", "ajax_info.txt", false);
xhttp.send();
document.getElementById("demo").innerHTML = xhttp.responseText;
```

# RISPOSTE XMLHttp

La responseTextproprietà restituisce la risposta del server come stringa JavaScript e puoi utilizzarla di conseguenza:

```python
document.getElementById("demo").innerHTML = xhttp.responseText;
```

**LA PROPRIETA' responseXML**

L'oggetto XMLHttpRequest ha un parser XML integrato.

La responseXMLproprietà restituisce la risposta del server come oggetto XML DOM.

Utilizzando questa proprietà è possibile analizzare la risposta come un oggetto XMLDOM:

```python
const xmlDoc = xhttp.responseXML;
const x = xmlDoc.getElementsByTagName("ARTIST");

let txt = "";
for (let i = 0; i < x.length; i++) {
  txt += x[i].childNodes[0].nodeValue + "<br>";
}
document.getElementById("demo").innerHTML = txt;

xhttp.open("GET", "cd_catalog.xml");
xhttp.send();
```

**METODI DI RISPOSTA DEL SERVER**

Il getAllResponseHeaders()metodo restituisce tutte le informazioni dell'intestazione dalla risposta del server.

```python
const xhttp = new XMLHttpRequest();
xhttp.onload = function() {
    document.getElementById("demo").innerHTML =
    this.getAllResponseHeaders();
}
xhttp.open("GET", "ajax_info.txt");
xhttp.send();
```

Il getResponseHeader()metodo restituisce informazioni specifiche dell'intestazione dalla risposta del server.

```python
const xhttp = new XMLHttpRequest();
xhttp.onload = function() {
    document.getElementById("demo").innerHTML =
    this.getResponseHeader("Last-Modified");
}
xhttp.open("GET", "ajax_info.txt");
xhttp.send();
```
