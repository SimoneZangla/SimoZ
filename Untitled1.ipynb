API sta per "Application Programming Interface" ed è una raccolta di metodi e
funzioni
implementati in un sistema software in modo che sia possibile comunicare con altre applicazioni.
In altre parole, un'API consente l'utilizzo di un servizio (funzione o metodo) di una piattaforma a
un sistema esterno, affinché possa essere utilizzato in un'applicazione proprietaria. Tali "metodi e
funzioni" possono includere operazioni di creazione, lettura, aggiornamento ed eliminazione. Un esempio calzante può essere il sito Trivago, in quanto selezionando il luogo e la data di interesse e facendo clic su "Cerca", il sito interagisce con l'API di ciascun hotel, fornendo
risultati per le camere disponibili che soddisfano i tuoi criteri.

# Che cos'è REST?

Nel 2000, Roy Fielding ha introdotto il concetto di REST (Representational State Transfer)
definendolo come uno stile architetturale per la progettazione e la creazione di servizi Web, in
altre parole REST rappresenta un insieme di regole che definiscono come scambiare risorse in un
sistema distribuito
I servizi web che implementano queste regole sono definiti RESTful.
Vediamo quali sono queste regole:
1. **Client-server:** la comunicazione tra le applicazioni segue il paradigma c/s. Questo
disaccoppiamento garantisce che, purché sia stata concordata un'interfaccia, lo sviluppo di
client e server può essere eseguito in modo indipendente;
2. **Stateless:** Nessuno stato client è memorizzato sul server. Tutte le informazioni necessarie per
eseguire le operazioni sono contenute nelle richieste;
3. **Cacheable:** I servizi Web RESTful devono fornire funzionalità di memorizzazione nella cache. I
server possono indicare come e per quanto tempo memorizzare nella cache le risposte; I
client possono utilizzare le risposte memorizzate nella cache invece di contattare il server.
Poiché REST in genere utilizza HTTP, eredita tutte le proprietà di memorizzazione nella cache
offerte da HTTP.
4. **Interfaccia uniforme:** le risorse che il server espone devono essere identificate da URI;
5. **Sistema a strati:** Data l’architettura c/s, i client non sono consapevoli del server specifico con
cui interagiscono. Questa proprietà consente l'introduzione di server intermedi che possono,
ad esempio, gestire la sicurezza o offrire funzionalità di bilanciamento del carico;
6. **Codice su richiesta (opzionale):** Anche se fa parte dell'architettura REST, questo principio è
facoltativo. I server possono estendere le funzionalità dei client trasferendo codice eseguibile.
REST, per definizione, è indipendente da qualsiasi protocollo sottostante e non è
necessariamente collegato a http, tuttavia le implementazioni REST più comuni usano HTTP
come protocollo applicativo.

# Che cos'è un RESTfull WebService?

Quando si sono diffuse le API per applicazioni e piattaforme Web, è stato introdotto il termine
RESTFUL WEB SERVICE, che viene utilizzato per denominare quelle API implementate con il
protocollo HTTP che forniscono una mappatura tra i verbi HTTP (get, post, put, delete) e le
azioni CRUD (create, read, update, delete).

**API vs Web Service**

API e Web Service sono due sistemi di comunicazione tra applicazioni. La principale differenza è
che un'API funge da interfaccia tra due diverse applicazioni in modo che possano comunicare tra
loro mentre un Web service ha lo stesso compito ma le applicazioni che mette in comunicazione
sono in rete. Un’API può utilizzare qualsiasi mezzo di comunicazione per avviare l'interazione tra
le applicazioni (es. API del sistema operativo sono utilizzabili chiamando delle funzioni in un
programma software). Un servizio Web usa http.
Tutti i WebService sono API ma tutte le API non sono WebService.
Le api che seguono i principi di rest e lavorano su http sono chiamate Restful API o Restful web
services

Uno dei vantaggi principali di REST su HTTP è costituito dal fatto che usa standard aperti e non
vincola l'implementazione dell'API o le applicazioni client a un'implementazione
specifica. Qualsiasi client deve poter chiamare l'API indipendentemente da come è implementata
internamente. Ad esempio, un servizio Web RESTful scritto in PHP, può essere consumato da
un’applicazione client scritta in un qualsiasi linguaggio (Java ad esempio) che può generare
richieste HTTP e analizzare risposte HTTP.
Vediamo i principali principi di progettazione di restful api che usano HTTP:


*   Le api sono progettate in base a risorse, che possono essere costituite da
qualsiasi
tipologia di oggetto, servizio o dato accessibile dal client.
*   Una risorsa ha un identificatore, costituito da un URI che identifica in modo univoco la
risorsa stessa.
*   I client interagiscono con un servizio scambiando rappresentazioni delle risorse. Molte
API usano JSON come formato di scambio.
*   Le API usano un'interfaccia uniforme che consente di separare le implementazioni del
client e del servizio. Per le API basate su HTTP, l'interfaccia uniforme include l'uso di verbi
HTTP standard per l'esecuzione di operazioni sulle risorse. Le operazioni più comuni sono
GET, POST, PUT, PATCH e DELETE.


*   Le API usano un modello di richiesta senza stato. Le richieste HTTP devono essere
indipendenti e possono essere eseguite in qualsiasi ordine, di conseguenza non è
possibile mantenere informazioni temporanee sullo stato tra le richieste. L'unico punto in
cui sono memorizzate le informazioni sono le risorse stesse e ogni richiesta deve essere
un'operazione atomica. Questo vincolo garantisce ai servizi Web un'elevata scalabilità,
perché non è necessario mantenere alcuna affinità tra i client e server specifici. Qualsiasi
server può gestire qualsiasi richiesta proveniente da qualsiasi client.
*   Le API REST si basano su collegamenti ipermediali contenuti nella rappresentazione
(HATEAOS*). Ad esempio nella rappresentazione JSON di un ordine sono contenuti i
collegamenti per recuperare o aggiornare il cliente associato all'ordine.
